##Gradio and Transformer

By using the pre trained model by using hugging AI

torch → Used for deep learning computations.

```python
import gradio as gr
from transformers import BlipProcessor,BlipForConditionalGeneration
from PIL import Image
```

```python
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")
```

```python
def genrate_caption(image):
    img=Image.open(image).convert("RGB")
    inputs = processor(img,return_tensors="pt")
    caption_ids = model.generate(**inputs)
    caption=processor.decode(caption_ids[0],skip_special_tokens=True)
    return caption
```

```python
interface = gr.Interface(
    fn=genrate_caption,
    inputs=gr.Image(type="filepath"),
    outputs="text",
    title="image caption genrator",
    description="upload an image and get a ai genrated caption"
)
```

```python
interface.launch()
```
